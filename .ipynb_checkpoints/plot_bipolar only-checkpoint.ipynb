```python
## 6 Hz
key = '6_Hz'
name = simulation['simulations'][key]['name']
stimulus = simulation['simulations'][key]['stimulus']
time = simulation['simulations'][key]['time']
lastflash_tp = simulation['simulations'][key]['lastflash_tp']
time = time - lastflash_tp
sol = simulation['simulations'][key]['sol']
RG = simulation['simulations'][key]['RG']
OPL_inputs = simulation['simulations'][key]['OPL_inputs']



ax0.plot(time,stimulus, color = 'grey')


#ax1.plot(time,sol.y[0],color = 'g', label = 'E_ON')
#ax1.plot(time,sol.y[2],color = 'orange', label = 'I_ON')
ax1.legend()



ax2.plot(time,sol.y[0],color = 'r',linestyle = ':', label = 'E_ON fixed')
ax2.plot(time,sol.y[0]*sol.y[1],color = 'r', label = 'E_ON dynamic')
ax2.legend()


ax3.plot(time,sol.y[1],color = 'c', label = 'n')

ax4.plot(time,sol.y[2], color = 'k', label = 'VG')



ax5.plot(time,RG, color = 'k', label = 'RG')

ax0.set_xticks([])
ax1.set_xticks([])
ax2.set_xticks([])
ax3.set_xticks([])
ax4.set_xticks([])


ax0.set_title(f'Stimulus {key}')
#ax1.set_title('ON Reponses ')
ax2.set_title('ON Response ')
ax3.set_title('Occupancy')
ax4.set_title('VG response')
ax5.set_title('RG response')
ax5.set_xlabel('time [s]')



### 16 Hz
key = '16_Hz'
name = simulation['simulations'][key]['name']
stimulus = simulation['simulations'][key]['stimulus']
time = simulation['simulations'][key]['time']
lastflash_tp = simulation['simulations'][key]['lastflash_tp']
time = time - lastflash_tp
sol = simulation['simulations'][key]['sol']
RG = simulation['simulations'][key]['RG']
OPL_inputs = simulation['simulations'][key]['OPL_inputs']



ax00.plot(time,stimulus, color = 'grey')


# ax11.plot(time,sol.y[0],color = 'g', label = 'E_ON')
# ax11.plot(time,sol.y[2],color = 'orange', label = 'I_ON')
# ax11.legend()



ax22.plot(time,sol.y[0],color = 'r',linestyle = ':', label = 'E_ON fixed')
ax22.plot(time,sol.y[0]*sol.y[1],color = 'r', label = 'E_ON dynamic')
ax22.legend()


ax33.plot(time,sol.y[1],color = 'c', label = 'n')

ax44.plot(time,sol.y[2], color = 'k', label = 'VG')



ax55.plot(time,RG, color = 'k', label = 'RG')

ax00.set_xticks([])
ax11.set_xticks([])
ax22.set_xticks([])
ax33.set_xticks([])
ax44.set_xticks([])


ax00.set_title(f'Stimulus {key}')
#ax11.set_title('ON Reponses ')
ax22.set_title('ON Response ')
ax33.set_title('Occupancy')
ax44.set_title('VG response')
ax55.set_title('RG response')
ax55.set_xlabel('time [s]')


ax00.set_ylabel('s(t)')
ax11.set_ylabel('V(t)')
ax22.set_ylabel('V(t)')
ax33.set_ylabel('n(t)')
ax44.set_ylabel('V(t)')
ax55.set_ylabel('R(t)')



```
