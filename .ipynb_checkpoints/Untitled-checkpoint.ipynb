```python
import cma
import numpy as np 
from scipy.ndimage import gaussian_filter
from BA1A2_Model import BA1A2_Model
from filter import filter_alpha_norm
from convolutions import convolve_1D
from dynamical_systems import IPL_rectified, IPL_rectified_occupancy, IPL_rectified_occupancy_nV,IPL_occupancy,IPL
from nonlinearities import N,sig
from utils import  make_param_dict, save_dict,save_fig,make_directory
from stimuli import impulse_stimulus, step_stimulus
from load_data import get_euler_stimulus,get_euler_response,get_OSR_response
from simulate import simulate_OSR
from visualizations import plot_OSR_simulation, plot_slope
from matplotlib.backends import backend_pdf
from matplotlib import pyplot as plt
import os
import numpy as np
import json



# compute error
def compute_mse(y,prediction):
        
    error = np.sum((prediction-y)**2)
    try:
        error /= len(y)
    except: 
        None
    error = np.sqrt(error)
    return error

# compute error
def compute_ae(y,prediction):


    error = 0
    for i,yi in enumerate(y): 
        error += np.abs(prediction[i]-yi)
    try:
        error /= len(y)
    except: 
        error = error
    return error




def optimize_to_chirp(Model_class,
                    x_init,
                    x_fixed,
                    param_names,
                    scale_values,
                    stim,
                    time,
                    true_response,
                    linear_filter,      
                    convolution,
                    system,
                    nonlinearity,
                    occupancy,
                    convolution_type,
                    polarities,
                    filterlength,
                    dt,
                    nb_repeats = 10, popsize = 10, sigma0 = 0.1):


    '''
    optimize all model parameter to best predict the chirp response

    with occupancy 


    '''
    # make sure all initial values are of similar orders of magnitude 


    # transform initial conditions into log scale
    x0 =  np.log(x_init)
    #x0 = x_init

    #start evolutionary strategy, multivatiate gaussian distributions based on x0 and sigma0
    es = cma.CMAEvolutionStrategy(x0, sigma0, {'popsize' : popsize})
    
    errors_all = []

    # loop over repeats 
    for n in range(nb_repeats):

        paramsets = es.ask()           # draw paramsets
        errors = []

        # loop over each parameterset drawn from the distribution
        for pset in paramsets:
                          
            #x = np.exp(pset) * scale_values   # mulitply by scaling factors to get params
            x = np.exp(pset)
            x = np.append(x,x_fixed)
            x = x * scale_values   # mulitply by scaling factors to get params

            params_init = {param_names[i] : x[i] for i in range(len(param_names))}

            params_model = make_param_dict(params_init)


            #initalize the model with the parameter 
            Model = Model_class(linear_filter = linear_filter,      
                            convolution = convolution,
                            system  = system,
                            nonlinearity = nonlinearity,
                            occupancy = occupancy,
                            convolution_type = convolution_type,
                            params = params_model,
                            polarities = polarities,
                            filterlength = filterlength,
                            dt = dt)



            #simulate the response
            simulation = Model.predict(stim,
                                       time,
                                        '_',
                                        print_steps = False)
            
            prediction = np.array(simulation['RG'])

            #calculate the error
            err = compute_mse(true_response,prediction)
            #err = compute_ae(true_response,prediction)
            
          
            errors.append(err)
                

        # evalutate all parametersets in the repeat 
        errors = np.asarray(errors)
        es.tell(paramsets,errors)
        es.disp()
        
        # save errors for each repetition
        errors_all.append(errors.mean())
    

    return es,errors_all,simulation




def optimize_linear_to_responses(Model_class,
                    x_init,
                    x_fixed,
                    param_names,
                    scale_values,
                    stim_euler,
                    time_euler,
                    true_response_euler,
                    true_response_osr,
                    linear_filter,      
                    convolution,
                    system,
                    nonlinearity,
                    occupancy,
                    convolution_type,
                    polarities,
                    filterlength,
                    dt,
                    nb_repeats = 10, popsize = 10, sigma0 = 0.1):


    '''
    optimize all model parameter to best predict the chirp response

    with occupancy 


    '''
    # make sure all initial values are of similar orders of magnitude 


    # transform initial conditions into log scale
    x0 =  np.log(x_init)
    #x0 = x_init

    #start evolutionary strategy, multivatiate gaussian distributions based on x0 and sigma0
    es = cma.CMAEvolutionStrategy(x0, sigma0, {'popsize' : popsize})
    
    errors_all = []

    # loop over repeats 
    for n in range(nb_repeats):

        paramsets = es.ask()           # draw paramsets
        errors = []

        # loop over each parameterset drawn from the distribution
        for pset in paramsets:
                          
            #x = np.exp(pset) * scale_values   # mulitply by scaling factors to get params
            x = np.exp(pset)
            x = np.append(x,x_fixed)
            x = x * scale_values   # mulitply by scaling factors to get params

            params_init = {param_names[i] : x[i] for i in range(len(param_names))}

            params_model = make_param_dict(params_init)


            #initalize the model with the parameter 
            Model = Model_class(linear_filter = linear_filter,      
                            convolution = convolution,
                            system  = system,
                            nonlinearity = nonlinearity,
                            occupancy = occupancy,
                            convolution_type = convolution_type,
                            params = params_model,
                            polarities = polarities,
                            filterlength = filterlength,
                            dt = dt)



            #simulate the response
            simulation_euler = Model.predict(stim_euler,
                                       time_euler,
                                        '_',
                                        print_steps = False)
            
            prediction_euler = np.array(simulation_euler['RG'])

            #calculate the error
            err_euler = compute_mse(true_response_euler,prediction_euler)
            
            
            #simulate the response
            osr_simulation = simulate_OSR(Model,
                                          osrparams,
                                          hyperparameter,
                                          params_model,
                                          dt = dt,
                                          show = False,
                                          print_steps = False)
            
            
            err_osr = 0 
            for i,key in enumerate(osr_simulation['simulations']):
                
                prediction_osr = osr_simulation['simulations'][key]['RG']
                #calculate the error
                err_osr =err_osr + compute_mse(true_response_osr[i],prediction_osr)

            
          
            errors.append(err_euler + 2*err_osr)
                

        # evalutate all parametersets in the repeat 
        errors = np.asarray(errors)
        es.tell(paramsets,errors)
        es.disp()
        
        # save errors for each repetition
        errors_all.append(errors.mean())
    

    return es,errors_all,simulation_euler,osr_simulation



def optimize_weights_array(Model,
                    x_init,
                    best_params,
                    true_delays,
                    true_response_osr,
                    slope_data,
                    dt,
                    osrparams,
                    hyperparameter,
                    nb_repeats = 30, popsize = 30, sigma0 = 0.1):


    '''
    optimize an array of weights for A1 to best match the OSR timing 

    '''
    
    wA1_init = best_params['w_A1']
    
    # make sure all initial values are of similar orders of magnitude 
    # transform initial conditions into log scale
    x0 =  np.log(x_init)
    #x0 = x_init

    #start evolutionary strategy, multivatiate gaussian distributions based on x0 and sigma0
    es = cma.CMAEvolutionStrategy(x0, sigma0, {'popsize' : popsize})
    
    errors_all = []
    
    # loop over repeats 
    for n in range(nb_repeats):

        paramsets = es.ask()           # draw paramsets
        errors = []

        # loop over each parameterset drawn from the distribution
        for pset in paramsets:
                          
            #x = np.exp(pset) * scale_values   # mulitply by scaling factors to get params
            x = np.exp(pset)
                    
            best_params['w_A1'] = wA1_init
            
            #simulate the response
            osr_simulation = simulate_OSR(Model,
                                          osrparams,
                                          hyperparameter,
                                          best_params,
                                          dt = dt,
                                          show = False,
                                          modified_weights = x,
                                          print_steps = False)
            

            
            delay_prediction = np.asarray(osr_simulation['delays'])
            slope_prediction = osr_simulation['slope']
            print(f'slope = {slope_prediction}')


            #calculate the error
            err_delay = compute_mse(true_delays,delay_prediction)
            err_slope = compute_mse(slope_data,slope_prediction)
            
            
            err_osr = 0 
            for i,key in enumerate(osr_simulation['simulations']):
                
                prediction_osr = osr_simulation['simulations'][key]['RG']
                #calculate the error
                err_osr =err_osr + compute_mse(true_response_osr[i],prediction_osr)

            
            
            errors.append(err_osr + 5*err_delay + err_slope*10)
                

        # evalutate all parametersets in the repeat 
        errors = np.asarray(errors)
        es.tell(paramsets,errors)
        es.disp()
        
        # save errors for each repetition
        errors_all.append(errors.mean())
    

    return es,errors_all,osr_simulation







def optimize_full_to_responses(Model_class,
                    x_init,
                    x_fixed,
                    param_names,
                    scale_values,
                    stim_euler,
                    time_euler,
                    true_response_euler,
                    true_response_osr,
                    true_delays,
                    slope_data,
                    linear_filter,      
                    convolution,
                    system,
                    nonlinearity,
                    occupancy,
                    convolution_type,
                    polarities,
                    filterlength,
                    dt,
                    nb_repeats = 10, popsize = 10, sigma0 = 0.1):


    '''
    optimize all model parameter to best predict chirp, osr response,delays annd slope


    '''

    # transform initial conditions into log scale
    x0 =  np.log(x_init)
    #x0 = x_init

    #start evolutionary strategy, multivatiate gaussian distributions based on x0 and sigma0
    es = cma.CMAEvolutionStrategy(x0, sigma0, {'popsize' : popsize})
    
    errors_all = []

    # loop over repeats 
    for n in range(nb_repeats):

        paramsets = es.ask()           # draw paramsets
        errors = []

        # loop over each parameterset drawn from the distribution
        for pset in paramsets:
                          
            #x = np.exp(pset) * scale_values   # mulitply by scaling factors to get params
            x = np.exp(pset)
            x = np.append(x,x_fixed)
            x = x * scale_values   # mulitply by scaling factors to get params

            params_init = {param_names[i] : x[i] for i in range(len(param_names))}

            params_model = make_param_dict(params_init)


            #initalize the model with the parameter 
            Model = Model_class(linear_filter = linear_filter,      
                            convolution = convolution,
                            system  = system,
                            nonlinearity = nonlinearity,
                            occupancy = occupancy,
                            convolution_type = convolution_type,
                            params = params_model,
                            polarities = polarities,
                            filterlength = filterlength,
                            dt = dt)



            #simulate the response
            simulation_euler = Model.predict(stim_euler,
                                       time_euler,
                                        '_',
                                        print_steps = False)
            
            prediction_euler = np.array(simulation_euler['RG'])

            #calculate the error
            err_euler = compute_mse(true_response_euler,prediction_euler)
            
            
            #simulate the response
            osr_simulation = simulate_OSR(Model,
                                          osrparams,
                                          hyperparameter,
                                          params_model,
                                          dt = dt,
                                          show = False,
                                          
                                          print_steps = False)
            
            delay_prediction = np.asarray(osr_simulation['delays'])
            slope_prediction = osr_simulation['slope']
            #print(f'slope = {slope_prediction}')


            #calculate the error
            err_delay = compute_mse(true_delays,delay_prediction)
            err_slope = compute_mse(slope_data,slope_prediction)
            
            
            err_osr = 0 
            for i,key in enumerate(osr_simulation['simulations']):
                
                prediction_osr = osr_simulation['simulations'][key]['RG']
                #calculate the error
                err_osr =err_osr + compute_mse(true_response_osr[i],prediction_osr)

            
          
            errors.append(err_euler +2*err_osr + 3*err_delay + 5*err_slope)
                

        # evalutate all parametersets in the repeat 
        errors = np.asarray(errors)
        es.tell(paramsets,errors)
        es.disp()
        
        # save errors for each repetition
        errors_all.append(errors.mean())
    

    return es,errors_all,simulation_euler,osr_simulation
```
