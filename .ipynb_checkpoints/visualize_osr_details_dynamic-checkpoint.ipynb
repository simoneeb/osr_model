# Model Responses at each stage

## Plot OPL outputs
 

```python
%matplotlib inline
fig, ax = plt.subplots(1,3, figsize = (20,5), sharey = True, sharex = True)

for key in simulation_dict_dynamic['simulations']:

    time = simulation_dict_dynamic['simulations'][key]["time"]
    
    for i,x in enumerate(simulation_dict_dynamic['simulations'][key]['OPL_inputs']):

        fun = simulation_dict_dynamic['simulations'][key]['OPL_inputs'][x]
        
        ax[i].plot(time, [fun(t) for t in time], label = key)
        ax[i].set_title(x)
        ax[i].set_xlabel('time [s]')

    ax[0].legend()
    

```

## Plot cell responses

```python
nb_cells = 3
names = ['VB', "VA1", 'VA2']

fig, ax = plt.subplots(1,3, figsize = (20,5), sharey = True, sharex = True)

for key in simulation_dict_dynamic['simulations']:

    time = simulation_dict_dynamic['simulations'][key]["time"]

    sol = simulation_dict_dynamic['simulations'][key]['sol']
    
    for i in range(nb_cells):
        
        ax[i].plot(time,sol.y[i], label = key)
        ax[i].set_title(names[i])
        ax[i].set_xlabel('time [s]')

    ax[0].legend()
    
```

## Plot occupancy

```python
fig = plt.figure()

for key in simulation_dict_dynamic['simulations']:

    time = simulation_dict_dynamic['simulations'][key]["time"]

    sol = simulation_dict_dynamic['simulations'][key]['sol']
    
    plt.plot(time,sol.y[3], label = key)
plt.title('nA1')
plt.xlabel('time [s]')
```

```python
%matplotlib qt
fig,ax = plt.subplots(7,sharex = True)


fig.subplots_adjust(top=0.88,
bottom=0.125,
left=0.735,
right=0.9,
hspace=0.2,
wspace=0.2)

#TODO plot plot the currents such that they start at 0 



time = simulation_dict_dynamic['simulations'][key]['time']
lastflashend = simulation_dict_dynamic['simulations'][key]['lastflash_tp']

time = time-lastflashend 

stim = simulation_dict_dynamic['simulations'][key]['stimulus']
sol = simulation_dict_dynamic['simulations'][key]['sol']
sol_2 = simulation_dict_2['simulations'][key]['sol']

delay_2 = simulation_dict_2['simulations'][key]["delay"]
amp_2 = simulation_dict_2['simulations'][key]["peak_amplitude"]


delay = simulation_dict_dynamic['simulations'][key]["delay"]
amp = simulation_dict_dynamic['simulations'][key]["peak_amplitude"]


VB = sol.y[0]
VA1 = sol.y[1]
VA2 = sol.y[2]
nA1 = sol.y[3]

VG = sol.y[-1]
VG_2 = sol_2.y[-1]

IB =  VB*w_B 
IA1 = VA1*nA1*w_A1
IA1_fixed = VA1*n_A1_star*w_A1
IA2 = VA2*w_A2


RG = simulation_dict_dynamic['simulations'][key]['RG'] 
RG_2 = simulation_dict_2['simulations'][key]['RG'] 


flashes = np.where(np.diff(stim) != 0)[0].tolist()

for i in range(len(flashes)):  
    if  not i % 2 :
        ax[0].axvspan(time[flashes[i]], time[flashes[i+1]+1], color = 'k', alpha = 0.1)
        ax[1].axvspan(time[flashes[i]], time[flashes[i+1]+1], color = 'k', alpha = 0.1)
        ax[2].axvspan(time[flashes[i]], time[flashes[i+1]+1], color = 'k', alpha = 0.1)
        ax[3].axvspan(time[flashes[i]], time[flashes[i+1]+1], color = 'k', alpha = 0.1)
        ax[4].axvspan(time[flashes[i]], time[flashes[i+1]+1], color = 'k', alpha = 0.1)
        ax[5].axvspan(time[flashes[i]], time[flashes[i+1]+1], color = 'k', alpha = 0.1)
        ax[6].axvspan(time[flashes[i]], time[flashes[i+1]+1], color = 'k', alpha = 0.1)




ax[0].set_title('Stimulus')
ax[1].set_title('ON Pathway')
ax[2].set_title('Amacrine Cell')
ax[3].set_title('Amacrine Cell Occupancy')
ax[4].set_title('Current Inputs')
ax[5].set_title('Ganlion Cell Volage')
ax[6].set_title('Firing Rate')

ax[0].set_ylabel('$stimulus intensity$')
ax[1].set_ylabel('$V$')
ax[2].set_ylabel('$V$')
ax[3].set_ylabel('$n$')
ax[4].set_ylabel('$I$')
ax[5].set_ylabel('$V$')
ax[6].set_ylabel('$R$')

ax[-1].set_xlabel('time [s]')


ax[0].set_title(key)
ax[0].plot(time,stim, label = '$s(t)$')
ax[1].plot(time,VB, color = 'blue', label = '$E_{1}$')
ax[1].plot(time,VA2, color = 'darkred', label = '$I_{1}$' )
ax[2].plot(time,VA1, color = 'red', label = '$I_{2}$')
ax[3].plot(time,nA1, color = 'c', label = '$n_{I2}$')


ax[4].plot(time,IB - IB[0], color = 'blue', label = '$E_{1}$')
ax[4].plot(time,IA1 - IA1[100], color = 'red', label = '$I_{1}$')
ax[4].plot(time,IA2 - IA2[100], color = 'darkred', label = '$I_{2}$')
#ax[4].plot(time,IA1_fixed - IA1_fixed[0], linestyle = ':', color = 'red', label = 'I_{I}_{2}')

ax[5].plot(time,VG, color = 'black', label = 'ON input only')
ax[5].plot(time,VG_2, color = 'black', linestyle = ':',  label = 'with $I_{2}$')

ax[6].plot(time,RG, color = 'black')
ax[6].plot(time,RG_2, color = 'black', linestyle = ':')

    

ax[6].scatter(delay,amp, color = 'black')
ax[6].scatter(delay_2,amp_2, color = 'black')

ax[6].axvline(delay, color = 'black')
ax[6].axvline(delay_2, color = 'black') 
ax[6].axvspan(delay,delay_2, color = 'black', alpha = 0.1)

ax[0].legend()
ax[1].legend()
ax[2].legend()
ax[3].legend()
ax[4].legend()
ax[5].legend()
ax[5].legend()


```
