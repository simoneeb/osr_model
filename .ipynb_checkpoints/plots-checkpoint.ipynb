```python
from filter import filter_alpha_norm
from utils import load_dict
import json
import numpy as np
import matplotlib.pyplot as plt
from nonlinearities import p
from visualizations import plot_OSR_3F
from visualizations import plot_delay_figure
import numpy as np 
```

```python
%matplotlib inline
```

# model parts

```python
simulation_name = 'params_manual_N_dynamic_OFF'
```

```python

# import params
filepath = f"/user/sebert/home/Documents/Repositories/osr_model/outputs/{simulation_name}"
with open(f'{filepath}/params.json') as json_file:
    params = json.load(json_file)
    print(params)
```

```python
filterlength = params['hyperparameter']['filterlength']
dt = params['hyperparameter']['dt']

filtertime = np.arange(0,filterlength,dt)

tau_B = params['params_model']['tau_B']
tau_A1 = params['params_model']['tau_A1']
tau_A2 = params['params_model']['tau_A2']
```

## plot filter

```python
fig = plt.figure()

fig.subplots_adjust(
top=0.88,
bottom=0.175,
left=0.19,
right=0.9,
hspace=0.2,
wspace=0.2)

ax = fig.add_subplot()
ax.set_xlabel("time [s]")
ax.locator_params(axis = 'x', nbins=2)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_yticks([])
#races.tick_params(axis='both', which='major', labelsize=fontsize_ticks)
ax.plot(filtertime,filter_alpha_norm(filtertime,tau_A2), color = 'darkred', linewidth = 5)
```

## plot nonlinearity

```python
xrange = np.arange(-1,1,0.02)


fig = plt.figure()

fig.subplots_adjust(
top=0.88,
bottom=0.175,
left=0.19,
right=0.9,
hspace=0.2,
wspace=0.2)

ax = fig.add_subplot()
ax.set_xlabel("input")
ax.set_ylabel("output")
ax.locator_params(axis = 'x', nbins=2)
ax.locator_params(axis = 'y', nbins=2)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.plot(xrange,[p(x,-0.5) for x in xrange], color = 'black', linewidth = 5)
```

## plot impulse response

```python
#load response
simulation_impulse = load_dict(f'{filepath}/simulation_impulse_dynamic.pickle')
simulation_impulse.keys()
```

```python
time = simulation_impulse['time']
response = simulation_impulse['sol'].y[-1]

fig = plt.figure()

fig.subplots_adjust(
top=0.88,
bottom=0.175,
left=0.19,
right=0.9,
hspace=0.2,
wspace=0.2)

ax = fig.add_subplot()
ax.set_xlabel('time')
ax.set_ylabel("$V_{G}$")
ax.set_xlim(1,2)
ax.locator_params(axis = 'x', nbins=2)
ax.locator_params(axis = 'y', nbins=2)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.plot(time,response, color = 'black', linewidth = 5)
```
