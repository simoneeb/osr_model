```python
from stimuli import impulse_stimulus
from stimuli import step_stimulus
from stimuli import periodic_flashes
from stimuli import periodic_flashes_mixed_polarities
from stimuli import periodic_flashes_variable_flashlength
from stimuli import get_euler_stimulus


import  matplotlib.pyplot as plt
from scipy.integrate import quad
import numpy as np
```

```python
#params 

dt = 0.002            # seconds
stimulus_length = 2  # seconds
```

# Stimuli with amplitude 1

```python
# test impulse
impulse,timeline_impulse = impulse_stimulus(stimulus_length,1,dt = dt)

impulse_itgr = np.cumsum(np.abs(impulse))[-1]*dt

plt.figure(figsize = (10,5))
plt.plot(timeline_impulse, impulse)
plt.xlabel('time [s]')
plt.title("Impulse Stimulus")


print(len(impulse), len(impulse)*dt)
print(f"Integral = {impulse_itgr}")
```

```python
# test flashes
freq = 20
fn = 12

flashes, timeline_flashes,tp_lastflash = periodic_flashes(freq,fn,dt = dt)

flashes_itgr = np.cumsum(np.abs(flashes))[-1]*dt


plt.figure(figsize = (10,5))
plt.plot(timeline_flashes, flashes)
plt.axvline(tp_lastflash)
plt.xlabel('time [s]')
plt.title("Periodic Flashes")


print(len(flashes), len(flashes)*dt)
print(f"Integral = {flashes_itgr}")

#TODO make sure the frequency is correct
```

```python
def periodic_flashes_mixed_polarities(frequency, flashnumber, dt = 0.02, delay = 2, flashduration = 0.04, polarities = -1*np.ones(12), baseline = 0) :
    
    period = (1/frequency) /dt
    flashduration = flashduration  /dt
    delay = delay /dt

    period = np.round(period)
    flashduration = np.round(flashduration)
    delay = np.round(delay)


    length =int( period * flashnumber + 2 * delay)
    timeline = np.arange(0,length)*dt
    
    stimulus = np.ones(length) * baseline
    tp_startflashes = int(delay)
    tp_endflashes = tp_startflashes + int(flashnumber * period)  - int(1)


    i=-1
    x = -1
    flash = 0
    for tp in range (tp_startflashes, tp_endflashes) :
        i=i+1
        if i% (period) < (flashduration):
            x = x +1
            print(f'i = {i}')
            print(f'x = {x}')
            
                
        if x != i:
            flash = flash+1
            x = i
            
        stimulus[tp] = polarities[flash]
        
        else:
            stimulus[tp] = baseline
        

    
    lastflashend = timeline[np.where(np.diff(stimulus) == polarities[-1])[0][-1]+int(1)]


    return np.asarray(stimulus),timeline,lastflashend
```

```python
# test flashes
freq = 20
fn = 12
flashpolarities = np.ones(fn)*-1
flashpolarities[-1] = 1
flashes, timeline_flashes,tp_lastflash = periodic_flashes_mixed_polarities(freq,fn,dt = dt,polarities = flashpolarities)

flashes_itgr = np.cumsum(np.abs(flashes))[-1]*dt


plt.figure(figsize = (10,5))
plt.plot(timeline_flashes, flashes)
plt.axvline(tp_lastflash)
plt.xlabel('time [s]')
plt.title("Periodic Flashes")


print(len(flashes), len(flashes)*dt)
print(f"Integral = {flashes_itgr}")

#TODO make sure the frequency is correct
```
