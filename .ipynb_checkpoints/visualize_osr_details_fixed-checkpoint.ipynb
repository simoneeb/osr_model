```python
from visualizations import plot_OSR_simulation,plot_slope
import json
from utils import load_dict
from matplotlib import pyplot as plt
from nonlinearities import p
import numpy as np
```

```python
# simulation to load
simulation_name = 'params_manual_N_fixed_noA1'
fn = 12
```

```python

# import params
filepath = f"/user/sebert/home/Documents/Repositories/osr_model/outputs/{simulation_name}"
with open(f'{filepath}/params.json') as json_file:
    params = json.load(json_file)
    print(params)
```

```python
# import simulations
simulation_dict = load_dict(f'{filepath}/simulation_osr_fixed_{fn}.pickle')
```

# Model Responses at each stage

## Plot OPL outputs
 

```python
fig, ax = plt.subplots(1,3, figsize = (20,5), sharey = True, sharex = True)

for key in simulation_dict['simulations']:

    time = simulation_dict['simulations'][key]["time"]
    
    for i,x in enumerate(simulation_dict['simulations'][key]['OPL_inputs']):

        fun = simulation_dict['simulations'][key]['OPL_inputs'][x]
        
        ax[i].plot(time, [fun(t) for t in time], label = key)
        ax[i].set_title(x)
        ax[i].set_xlabel('time [s]')

    ax[0].legend()
    

```

## Plot cell responses

```python
nb_cells = 3
names = ['VB', "VA1", 'VA2']

fig, ax = plt.subplots(1,3, figsize = (20,5), sharey = True, sharex = True)

for key in simulation_dict['simulations']:

    time = simulation_dict['simulations'][key]["time"]

    sol = simulation_dict['simulations'][key]['sol']
    
    for i in range(nb_cells):
        
        ax[i].plot(time,sol.y[i], label = key)
        ax[i].set_title(names[i])
        ax[i].set_xlabel('time [s]')

    ax[0].legend()
    
```

## plot gangion cell response

```python
fig, ax = plt.subplots(2,1, figsize = (20,5) , sharex = True)

for key in simulation_dict['simulations']:

    time = simulation_dict['simulations'][key]["time"]

    sol = simulation_dict['simulations'][key]['sol']
    RG = simulation_dict['simulations'][key]['RG']
    
    ax[0].plot(time,sol.y[3], label = key)
    ax[1].plot(time,RG, label = key)

ax[0].set_title('VG')
ax[0].legend()
ax[1].set_title('RG')
ax[1].set_xlabel('time [s]')
```

# visualize how AC makes peak 


```python
fig, ax = plt.subplots(3,1, figsize = (5,5) , sharex = True)

key = '6_Hz'

time = simulation_dict['simulations'][key]['time']
stim = simulation_dict['simulations'][key]['stimulus']
sol = simulation_dict['simulations'][key]['sol']

ax[0].set_title('Stimulus')
ax[0].plot(time,stim)

ax[1].set_title('Intermediate cells')
ax[1].plot(time,sol.y[0], label = 'Bipolar')
ax[1].plot(time,sol.y[1], label = 'Amacrine')
ax[1].legend()



ax[2].set_title('Gaglion Cell')
ax[2].plot(time,sol.y[3])


ax[-1].set_xlabel('time [s]')

ax[1].set_xlim(1.9,5.0)
```

```python
theta_B = params['params_model']['theta_B']
theta_A1 = params['params_model']['theta_A1']
theta_A2 = params['params_model']['theta_A2']

w_B = params['params_model']['w_B']
w_A1 = params['params_model']['w_A1']
w_A2 = params['params_model']['w_A2']

n_A1_star = params['params_model']['n_A1_star']
```

```python
key = '6_Hz'

time = simulation_dict['simulations'][key]['time']
stim = simulation_dict['simulations'][key]['stimulus']
sol = simulation_dict['simulations'][key]['sol']


VB = sol.y[0]
VA1 = sol.y[1]
VA2 = sol.y[2]
VG = sol.y[3]

IB =  np.array([p(v,theta_B) for v in VB])*w_B
IA1 = np.array([p(v,theta_A1) for v in VA1])*n_A1_star*w_A1
IA2 = np.array([p(v,theta_A2) for v in VA2])*w_A2


RG = simulation_dict['simulations'][key]['RG'] 








```

```python
%matplotlib qt
fig,ax = plt.subplots(8,1,sharex = True)


ax[0].set_title('Stimulus')
ax[0].plot(time,stim)

ax[1].set_title('ON Bipolar Cell')
ax[1].plot(time,VB, color = 'blue')

ax[2].set_title('ON Amacrine Cell A2')
ax[2].plot(time,VA2, color = 'darkred')

ax[3].set_title('ON Amacrine Cell A1')
ax[3].plot(time,VA1, color = 'red')

ax[4].set_title('Currents')
ax[4].plot(time,IB, color = 'blue')
ax[4].plot(time,IA2, color = 'darkred')
ax[4].plot(time,IA1, color = 'red')



ax[5].set_title('Ganlion Cell')
ax[5].plot(time,VG, color = 'black')

ax[6].set_title('Firing Rate')
ax[6].plot(time,RG, color = 'black')





fig.suptitle('Response without A1')

```
