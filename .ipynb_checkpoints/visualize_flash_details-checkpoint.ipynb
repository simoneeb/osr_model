# Model Responses at each stage

## Plot OPL outputs
 

```python
fig, ax = plt.subplots(1,3, figsize = (20,5), sharey = True, sharex = True)

time = simulation_dict_dynamic['time']

for i,x in enumerate(simulation_dict_dynamic['OPL_inputs']):

    fun = simulation_dict_dynamic['OPL_inputs'][x]

    
    ax[i].plot(time, [fun(t) for t in time])
    ax[i].set_title(x)
    ax[i].set_xlabel('time [s]')

ax[0].legend()
    

```

## Plot cell responses

```python
nb_cells = 3
names = ['VB', "VA1", 'VA2']

fig, ax = plt.subplots(1,3, figsize = (20,5), sharey = True, sharex = True)


for i in range(nb_cells):

    res = simulation_dict_dynamic['sol'].y[i]

    
    ax[i].plot(time, res)
    ax[i].set_title(names[i])
    ax[i].set_xlabel('time [s]')

ax[0].legend()
    
```

## Plot occupancy

```python
fig = plt.figure()

nA1 = simulation_dict_dynamic['sol'].y[3]

plt.plot(time,nA1)
plt.title('nA1')
plt.xlabel('time [s]')

#plt.xlim(1,2)
#plt.ylim(0.5,.6)
```

## plot gangion cell response

```python
fig, ax = plt.subplots(2,1, figsize = (20,5) , sharex = True)


sol = simulation_dict_dynamic['sol']
RG = simulation_dict_dynamic['RG']

ax[0].plot(time,sol.y[4])
ax[1].plot(time,RG)

ax[0].set_title('VG')
ax[0].legend()
ax[1].set_title('RG')
ax[1].set_xlabel('time [s]')

#ax[1].set_xlim(1,2.5)
#ax[0].set_ylim(-0.005,0.005)
```

# visualize for paper: how rebound is created

```python
lw_fig = 10
lw = 10
fontsize = 40
fontsize_legend = 20
fontsize_ticks = 20

%matplotlib qt
```

```python
stim = simulation_dict_dynamic['stimulus']
sol = simulation_dict_dynamic['sol']
sol_strych = simulation_dict_dynamic_strych['sol']
rate = simulation_dict_dynamic['RG']
rate_strych = simulation_dict_dynamic_strych['RG']

fig = plt.figure()

fig.subplots_adjust(
top=0.875,
bottom=0.175,
left=0.46,
right=0.9,
hspace=0.2,
wspace=0.2)

ax = fig.add_subplot()
ax.set_xlabel('time', fontsize = fontsize, fontweight = 'bold' )
ax.set_ylabel("$V_{G}$", fontsize = fontsize, fontweight = 'bold' )
ax.set_xlim(0,2)
ax.locator_params(axis = 'x', nbins=2)
ax.locator_params(axis = 'y', nbins=2)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.tick_params(axis='both', which='minor', labelsize=fontsize_ticks)
ax.tick_params(axis='both', which='major', labelsize=fontsize_ticks)



#ax.plot(time,rate, color = 'black', linewidth = lw_fig)
ax.plot(time,rate, color = 'black', linewidth = lw_fig)
ax.plot(time,rate_strych, color = 'gray', linewidth = lw_fig)
ax.axvspan(*time[np.where(np.diff(stim) != 0)], color = 'k', alpha = 0.1)
```

```python
%matplotlib qt
fig, ax = plt.subplots(4,1, figsize = (5,5) , sharex = True)


fig.subplots_adjust(
top=0.875,
bottom=0.175,
left=0.46,
right=0.9,
hspace=0.2,
wspace=0.2)




ax[0].set_title('Stimulus', fontsize = fontsize, fontweight = 'bold')
ax[0].plot(time,stim, color = 'k', linewidth = lw_fig)
ax[0].tick_params(axis='both', which='minor', labelsize=fontsize_ticks)
ax[0].tick_params(axis='both', which='major', labelsize=fontsize_ticks)


ax[1].set_title('Intermediate cells', fontsize = fontsize, fontweight = 'bold')
ax[1].plot(time,sol.y[0], color = 'blue', linewidth = lw_fig, label = 'Bipolar')
ax[1].plot(time,sol.y[1], color = 'red', linewidth = lw_fig, label = 'Amacrine 1', linestyle = ':')
ax[1].plot(time,sol.y[2], color = 'darkred', linewidth = lw_fig, label = 'Amacrine 2')
#ax[1].plot(time,sol_strych.y[2], label = 'Amacrine 2 inhibited')
ax[1].legend(fontsize = fontsize_legend)

ax[1].tick_params(axis='both', which='minor', labelsize=fontsize_ticks)
ax[1].tick_params(axis='both', which='major', labelsize=fontsize_ticks)


ax[2].set_title('Gaglion Cell', fontsize = fontsize, fontweight = 'bold')
ax[2].plot(time,sol_strych.y[4], color = 'k', linewidth = lw_fig, label = 'without 3rd pathway')
ax[2].plot(time,sol.y[4], color = 'k', linestyle = ':',  linewidth = lw_fig, label = 'with 3rd pathway')
ax[2].legend(fontsize = fontsize_legend)

ax[2].tick_params(axis='both', which='minor', labelsize=fontsize_ticks)
ax[2].tick_params(axis='both', which='major', labelsize=fontsize_ticks)

ax[3].set_title('Firing Rate', fontsize = fontsize, fontweight = 'bold')
ax[3].plot(time,rate_strych, color = 'k', linewidth = lw_fig)
ax[3].plot(time,rate, color = 'k', linestyle = ':', linewidth = lw_fig)
ax[3].axvline(time[np.argmax(rate_strych)], color = 'k', linewidth = lw_fig)
ax[3].axvline(time[np.argmax(rate)], color = 'k',linestyle = ':', linewidth = lw_fig)
ax[-1].tick_params(axis='both', which='minor', labelsize=fontsize_ticks)
ax[-1].tick_params(axis='both', which='major', labelsize=fontsize_ticks)
ax[-1].set_xlabel('time [s]', fontsize = fontsize, fontweight = 'bold')

ax[1].set_xlim(0.9,2.0)
```
